```python
#import vector
import numpy as np
import math
from matplotlib import pyplot as plt

nx = 5
x_max = 1
dx = x_max / nx
ny = 5
y_max = 1
dy = y_max / ny

x_pos = np.linspace(0, x_max, nx)
y_pos = np.linspace(0, y_max, ny)
pressures = np.zeros((nx*ny - 1))

index = 0
for i,x in enumerate(x_pos):
    for j, y in enumerate(y_pos):
        if i ==0 and j==0:
            pass
        else:
            pressures[index] = x+y
            index += 1

# Unpack pressure array back into 2D coordinates
pressure_grid = np.empty((nx,ny))
index = 0

for i in range(0, nx):
    for j in range(0, ny):
        if i==0 and j ==0:
            pressure_grid[i, j] = 0
        else:
            pressure_grid[i, j] = pressures[index]
            index += 1

print(pressure_grid)

X, Y = np.meshgrid(x_pos, y_pos)

fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
surf = ax.plot_surface(X, Y, pressure_grid)
```
